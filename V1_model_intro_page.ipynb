# Biologically-realistic model of a mouse V1 cortical column

The notebook to run through the model and analysis from [Billeh et al. "Systematic Integration of Structural and Functional Data into Multi-scale Models of Mouse Primary Visual Cortex". *Neuron* (2020)](https://www.cell.com/neuron/fulltext/S0896-6273(20)30067-2?dgcid=raven_jbs_etoc_email) is available through our collaborators at Cybershuttle and Georgia Tech. Cybershuttle is a platform that enables resource-intensive computing using remote HPC resources. We can run the V1 model on a laptop through Cybershuttle, when it would otherwise be much more difficult to set up the software and run the model on a laptop.

Click this link -> [Cybershuttle homepage](https://cybershuttle.org/)



Go to "Login" in the upper right corner.

Under "Select an Identity Provider" select "Google"

Use a google email address to create an account on CyberShuttle.

Click "Log On"


On the main heading bar, click "catalog" and click the tag for "allenai" or search for "Bio-realistic multiscale simulation of cortical circuits". Click on this item.

Once you've logged on, you should also be able to access the notebook through this link:

[V1 model tutorial link](https://cybershuttle.org/resources/REPOSITORY/4e48c577-0110-4a16-be0b-1b1bb4162cdb)

Once you click on the item, it will take you to the next page:

On this page, click "Start Project Session" to begin a virtual session.



Click "Create session". It might take a few seconds, but eventually you will land on a JupyterLab environment:

Click on "V1_notebook.ipynb" in the sidebar to access the tutorial.
